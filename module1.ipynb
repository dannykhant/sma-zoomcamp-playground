```python
ytd_performance = {}

print("Calculating YTD performance...")
for country, data in index_data.items():
    if not data.empty:
        # Ensure 'Close' column exists
        if 'Close' in data.columns:
            first_close = data['Close'].iloc[0] # First available closing price
            last_close = data['Close'].iloc[-1]  # Last available closing price

            # Calculate YTD growth
            # Extract the scalar value from the Series before comparison
            if first_close.item() != 0:
                growth = ((last_close.item() - first_close.item()) / first_close.item()) * 100
                ytd_performance[country] = growth
            else:
                ytd_performance[country] = None # Handle cases where first_close is 0
                print(f"Warning: First closing price for {country} is zero. Cannot calculate growth.")
        else:
            ytd_performance[country] = None
            print(f"Warning: 'Close' column not found for {country}.")
    else:
        ytd_performance[country] = None
        print(f"Warning: No data available for {country}. Cannot calculate YTD performance.")

# Convert to a DataFrame for better presentation
ytd_df = pd.DataFrame(list(ytd_performance.items()), columns=['Country', 'YTD Growth (%)'])

# Sort by YTD Growth for easy comparison
ytd_df = ytd_df.sort_values(by='YTD Growth (%)', ascending=False).reset_index(drop=True)

print("\nYear-to-Date (YTD) Performance (Jan 1 - Aug 21, 2026):")
display(ytd_df)
```

```python
import yfinance as yf
import pandas as pd

# Download S&P 500 historical data from 1950 to present
ticker = '^GSPC' # S&P 500 ticker
start_date_q3 = '1950-01-01'
end_date_q3 = pd.Timestamp.now().strftime('%Y-%m-%d')

print(f"Downloading S&P 500 data from {start_date_q3} to {end_date_q3}...")
sp500_data = yf.download(ticker, start=start_date_q3, end=end_date_q3)

# Use 'Close' price for analysis
sp500_close = sp500_data['Close'].dropna()

print("S&P 500 data downloaded successfully.")
display(sp500_close.head())
```

```python
# Identify all-time high closing price points
all_time_highs = pd.Series(dtype='float64')
current_all_time_high = -1.0 # Initialize with a very low value to ensure the first price is considered a new high

# Iterate over the Series of prices (Date and scalar value) for the '^GSPC' column
# This assumes sp500_close is a DataFrame with a single column named '^GSPC'
for date, close_price in sp500_close['^GSPC'].items():
    if close_price > current_all_time_high:
        current_all_time_high = close_price
        all_time_highs.loc[date] = current_all_time_high

print("All-time high closing price points identified.")
display(all_time_highs.head())

# Prepare to find corrections (drawdowns)
corrections_data = []

# Convert all_time_highs index to datetime for easier comparison and slicing
all_time_highs.index = pd.to_datetime(all_time_highs.index)
sp500_close.index = pd.to_datetime(sp500_close.index)

# Iterate through consecutive all-time highs
for i in range(len(all_time_highs) - 1):
    ath1_date = all_time_highs.index[i]
    ath1_price = all_time_highs.iloc[i]
    ath2_date = all_time_highs.index[i+1]

    # Get the prices between the two all-time highs (exclusive of ath1_date, inclusive of ath2_date)
    # The minimum should be found between ath1_date and ath2_date. The end date for slicing needs to be inclusive.
    prices_between_highs = sp500_close['^GSPC'].loc[ath1_date:ath2_date]

    # The minimum price might be the second ATH itself if there's no drawdown, or it could be before the second ATH.
    # We are looking for the minimum *after* the first ATH and *up to or including* the second ATH.
    # However, for a 'correction', we are interested in the dip *before* a new ATH is established.
    # So we should look for the minimum between ath1_date (exclusive) and ath2_date (exclusive).
    # If the market immediately sets a new high, there's no dip.

    # Let's consider the period from the current ATH up to the next ATH's day - 1 to find the lowest point.
    # If the next ATH is the very next day, there's no period to drop.

    # To find the true correction, we need to consider prices between current ATH and the subsequent ATH.
    # The 'low' of a correction should be the minimum price *after* ath1_date and *before* or *at* ath2_date.

    # Filter prices from the day *after* the current all-time high up to the next all-time high
    if (ath2_date - ath1_date).days > 0: # Ensure there's a period between ATHs
        period_prices = sp500_close['^GSPC'].loc[ath1_date + pd.Timedelta(days=1):ath2_date]

        if not period_prices.empty:
            min_price_in_period = period_prices.min()
            min_price_date = period_prices.idxmin()

            # If the minimum price is lower than the current ATH
            if min_price_in_period < ath1_price:
                drawdown = ((ath1_price - min_price_in_period) / ath1_price) * 100
                duration = (min_price_date - ath1_date).days

                corrections_data.append({
                    'ATH_Start_Date': ath1_date,
                    'ATH_Start_Price': ath1_price,
                    'Min_Price_Date': min_price_date,
                    'Min_Price': min_price_in_period,
                    'Drawdown_Percent': drawdown,
                    'Duration_Days': duration
                })

# Create a DataFrame from the corrections data
corrections_df = pd.DataFrame(corrections_data)

print("\nCorrections analysis completed.")
# Filter for corrections of at least 5%
major_corrections_df = corrections_df[corrections_df['Drawdown_Percent'] >= 5.0].reset_index(drop=True)

print("\nMajor corrections (>= 5% drawdown):")
display(major_corrections_df.head())
print(f"Total major corrections found: {len(major_corrections_df)}")

# Calculate percentiles for Drawdown_Percent and Duration_Days
drawdown_percentiles = major_corrections_df['Drawdown_Percent'].quantile([0.25, 0.5, 0.75])
duration_percentiles = major_corrections_df['Duration_Days'].quantile([0.25, 0.5, 0.75])

print("\nDrawdown Percentiles (>= 5% corrections):")
print(drawdown_percentiles)

print("\nDuration Percentiles (days, >= 5% corrections):")
print(duration_percentiles)
```

```python
ticker = 'AMZN'
ticker_obj = yf.Ticker(ticker)
result = ticker_obj.get_earnings_dates()
```

```python
# Download historical price data for AMZN
amzn_data = yf.download(ticker, period='max')
amzn_close = amzn_data['Close'].dropna()

print(f"Downloaded historical data for {ticker}.")
display(amzn_close.head())
```

```python
# Calculate 2-day percentage changes
# The shift(2) means we are comparing the current day's close price with the close price from 2 trading days ago.
amzn_2day_returns = amzn_close.pct_change(periods=2).dropna()

print("2-day percentage changes calculated.")
display(amzn_2day_returns.head())
```

```python
# Clean up the earnings dates and link with 2-day returns

# Ensure 'Earnings Date' is datetime and convert to date only for merging
# The result DataFrame has a MultiIndex. Resetting it to access 'Earnings Date' column.
earnings_df = result.reset_index()
earnings_df['Earnings Date'] = earnings_df['Earnings Date'].dt.normalize()

# Remove timezone information from 'Earnings Date' to ensure compatibility with amzn_close.index
earnings_df['Earnings Date'] = earnings_df['Earnings Date'].dt.tz_localize(None)

# Rename columns for clarity
earnings_df.rename(columns={'Earnings Date': 'Date', 'Surprise(%)': 'Surprise_Percent'}, inplace=True)

# Filter for relevant columns
earnings_df = earnings_df[['Date', 'Surprise_Percent']]

# Ensure amzn_2day_returns index is also date only for merging
# amzn_close.index is already timezone-naive (datetime64[ns]), so no change needed there.
amzn_2day_returns.index = amzn_2day_returns.index.normalize()

# earning_dates_series is earnings_df['Date']
earning_dates_series = earnings_df['Date']

# Find the next trading day for each earnings date
next_trading_day = []
for ear_date in earning_dates_series:
    # Find all trading days after or on the earnings date
    # amzn_close.index is timezone-naive, ear_date is now also timezone-naive
    future_trading_days = amzn_close.index[amzn_close.index >= ear_date]
    if len(future_trading_days) >= 2: # Need at least Day 2 (earnings day) and Day 3 (return day)
        # The first day is the earnings day (Day 2).
        # The second day is the day after earnings (Day 3), which is the index for amzn_2day_returns.
        next_trading_day.append(future_trading_days[1]) # This is Day 3
    else:
        next_trading_day.append(pd.NaT)

earnings_df['Return_Date'] = next_trading_day

# Drop rows where we couldn't find a next trading day (e.g., last earnings date with no subsequent trading day for return calculation)
earnings_df.dropna(subset=['Return_Date'], inplace=True)

# Merge based on the calculated 'Return_Date' (Day 3) and `amzn_2day_returns` index
merged_df = pd.merge(earnings_df, amzn_2day_returns, left_on='Return_Date', right_index=True, how='inner')

# Rename the return column for clarity
merged_df.rename(columns={'AMZN': '2_Day_Return'}, inplace=True)

# Filter for positive earnings surprises
positive_surprise_df = merged_df[merged_df['Surprise_Percent'] > 0].copy()

print("Merged earnings and 2-day returns. Filtered for positive surprises.")
display(positive_surprise_df.head())

# Calculate the median 2-day return for positive surprises
median_2day_return = positive_surprise_df['2_Day_Return'].median()
print(f"\nMedian 2-day return for positive earnings surprises: {median_2day_return:.4f}")

# Calculate the correlation between 2-day stock return and earnings surprise magnitude
correlation = positive_surprise_df['2_Day_Return'].corr(positive_surprise_df['Surprise_Percent'])
print(f"\nCorrelation between 2-day stock return and positive earnings surprise magnitude: {correlation:.4f}")
```
